# El decorador `@property`

El decorador `@property` en Python sirve para hacer que un **método** de una clase se comporte
como un **atributo**. Es especialmente útil cuando quieres controlar cómo se obtiene o modifica
un valor sin cambiar la forma en que se utiliza desde fuera.

## 1. El problema que resuelve

Imagina una clase sencilla:

````python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre
````

Puedes hacer:

````python
persona = Persona("Ana")
print(persona.nombre)
````

Aquí `nombre` es simplemente un atributo.

Pero quizá quieras que `nombre` se calcule, se valide, o se transforme antes de devolverlo.
Podrías crear un método:

````python
class Persona:
    def __init__(self, nombre):
        self._nombre = nombre

    def obtener_nombre(self):
        return self._nombre
````

Entonces:

````python
persona = Persona("Ana")
print(persona.obtener_nombre())
````

Funciona, pero es menos natural. Ahí entra `@property`.

## 2. `@property` convierte un método en un atributo

Podemos escribir:

````python
class Persona:
    def __init__(self, nombre):
        self._nombre = nombre

    @property
    def nombre(self):
        return self._nombre
````

Ahora podemos hacer:

````python
persona = Persona("Ana")

print(persona.nombre)
````

Fíjate en algo importante:

```python
persona.nombre
```

y no:

```python
persona.nombre()
```

Aunque `nombre` está definido como un método:

```python
@property
def nombre(self):
    return self._nombre
```

Python hace que podamos utilizarlo como si fuera un atributo.

## 3. ¿Qué está pasando realmente?

Cuando escribes:

```python
@property
def nombre(self):
    return self._nombre
```

le estás diciendo a Python, aproximadamente:

> *"Cuando alguien intente acceder a `persona.nombre`, ejecuta este método y devuelve su resultado."*

Por eso:

```python
persona.nombre
```

termina ejecutando internamente:

```python
persona.nombre()
```

pero tú no escribes los paréntesis.

Es una forma de crear un **getter**.

## 4. ¿Por qué usarlo en lugar de un atributo normal?

Supongamos que tenemos:

````python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto
````

Podemos calcular el área con un método:

````python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def area(self):
        return self.ancho * self.alto
````

Y usarlo:

````python
rectangulo = Rectangulo(4, 5)
print(rectangulo.area())
````

Pero conceptualmente el área es una **propiedad** del rectángulo, no una **acción** que le
pedimos realizar.

Con `@property`:

````python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    @property
    def area(self):
        return self.ancho * self.alto
````

Ahora:

````python
rectangulo = Rectangulo(4, 5)
print(rectangulo.area)
````

Es muy cómodo porque parece un atributo:

```python
rectangulo.area
```

pero realmente Python está ejecutando código para obtenerlo.

## 5. Una propiedad es de solo lectura

Fíjate en un detalle: con `@property` hemos definido cómo se **lee** `area`, pero no cómo se
**escribe**. Por defecto, una propiedad definida así es de **solo lectura**: se puede
consultar, pero no asignar.


````python
class Rectangulo:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    @property
    def area(self):
        return self.ancho * self.alto
````

Puedes hacer:

````python
rectangulo = Rectangulo(4, 5)
print(rectangulo.area)
````

pero no `rectangulo.area = 100`. Python dará un error porque `area` es una propiedad de **solo lectura**:

````python
try:
    rectangulo.area = 100
except AttributeError as e:
    print("AttributeError:", e)
````

Esto es muy útil para valores que quieres calcular pero que no deberían poder modificarse
directamente.

## 6. (Opcional) `@property` también permite controlar la escritura

Esta parte es más avanzada y no hace falta para el ejercicio, pero merece la pena conocerla.

Puedes definir no solamente cómo **leer** una propiedad, sino también cómo **asignarle** un
valor. Para eso utilizas `@nombre.setter`.

Por ejemplo:

````python
class Persona:
    def __init__(self, edad):
        self._edad = edad

    @property
    def edad(self):
        return self._edad

    @edad.setter
    def edad(self, valor):
        if valor < 0:
            raise ValueError("La edad no puede ser negativa")

        self._edad = valor
````

Ahora:

````python
persona = Persona(25)

print(persona.edad)
````

Pero podemos hacer:

````python
persona.edad = 30
print(persona.edad)
````

y Python ejecutará el setter:

```python
@edad.setter
def edad(self, valor):
    ...
```

En cambio, `persona.edad = -10` provocará un `ValueError`:

````python
try:
    persona.edad = -10
except ValueError as e:
    print("ValueError:", e)
````
