## 1. ¿Qué es una PEP y qué es la PEP 8?

**PEP** son las siglas de *Python Enhancement Proposal* (Propuesta de Mejora de
Python): documentos donde se discuten y deciden cambios y convenciones del lenguaje.

La **PEP 8** es la **guía de estilo** para escribir código Python. No cambia lo que el
programa *hace* (al intérprete le da igual el estilo), sino cómo se *ve*. Su objetivo
es uno: la **consistencia**, que hace el código más legible.

> 💡 Una frase de la propia PEP 8 que conviene recordar: *"la legibilidad cuenta"*, y
> *"la consistencia con esta guía es importante, pero la consistencia dentro de un
> proyecto lo es más"*. Es decir: las reglas son una guía, no un dogma.

---

## 2. Nombres (la convención que más se nota)

Elegir bien los nombres y seguir la convención de mayúsculas/minúsculas correcta es lo
que más distingue al código "de Python". La regla según *qué* nombramos:

| Elemento | Convención | Ejemplo |
|----------|------------|---------|
| Variables y funciones | `snake_case` (minúsculas con guion bajo) | `numero_intentos`, `calcular_total()` |
| Clases | `PascalCase` (cada palabra en mayúscula) | `CuentaBancaria`, `MetodoPago` |
| Constantes | `MAYUSCULAS_CON_GUION` | `NUMERO_MAXIMO`, `PI` |
| Módulos / archivos | `minusculas` (cortas, sin guiones medios) | `utilidades.py` |
| "Privado" (uso interno) | Prefijo con un guion bajo | `_contador`, `_ayudante()` |

Veámoslo todo junto en un ejemplo:

````python
NUMERO_MAXIMO_INTENTOS = 3  # constante -> MAYÚSCULAS


class CuentaBancaria:  # clase -> PascalCase
    def __init__(self, titular_cuenta):
        self.titular_cuenta = titular_cuenta  # variable -> snake_case


def calcular_interes(saldo, tasa):  # función -> snake_case
    return saldo * tasa


cuenta = CuentaBancaria("Ana")
print(cuenta.titular_cuenta)
print(calcular_interes(1000, 0.02))
````

> 💡 Y, por encima de la convención de mayúsculas: usa **nombres descriptivos**.
> `dias_restantes` se entiende; `d` o `x`, no. Evita abreviaturas crípticas y nombres
> de una sola letra (salvo en bucles muy cortos, como `for i in range(...)`).

## 3. Indentación y disposición del código

- **Indentación: 4 espacios** por nivel. Nada de tabuladores, y nunca mezclar tabs y
  espacios (Python da error si los mezclas).
- **Longitud de línea**: la PEP 8 recomienda un máximo de **79 caracteres**. Las
  herramientas modernas (Black, Ruff) usan **88** por defecto, que es lo habitual hoy.
  Lo importante es **fijar un límite** y respetarlo en todo el proyecto.
- **Líneas en blanco**:
  - **2 líneas** en blanco entre funciones o clases de nivel superior.
  - **1 línea** en blanco entre los métodos de una clase.
  - Líneas en blanco sueltas dentro de una función para separar bloques lógicos.

```python
import os


class Usuario:                 # 2 líneas en blanco antes de una clase
    def __init__(self, nombre):
        self.nombre = nombre

    def saludar(self):         # 1 línea en blanco entre métodos
        return f"Hola, {self.nombre}"


def funcion_suelta():          # 2 líneas en blanco antes de una función top-level
    return 42
```

> 💡 No te obsesiones con contar líneas y espacios a mano: para esto sirve
> precisamente el **formateador de Ruff** (`ruff format`), que reorganiza todo esto por
> ti. Conviene conocer las reglas, pero la máquina las aplica.

---

## 4. Espacios en blanco alrededor de operadores y signos

Los espacios bien puestos hacen el código mucho más legible. Las reglas básicas:

- **Un espacio** alrededor de los operadores (`=`, `+`, `==`, `<`...) y **después** de
  las comas.
- **Sin espacios** justo por dentro de paréntesis, corchetes o llaves.
- **Sin espacios** alrededor del `=` cuando es un **argumento con valor por defecto**
  o se pasa por nombre (`funcion(x=1)`).

Compara:

```python
# ⚠️ Mal
total=precio+ iva*cantidad
lista = [ 1,2 ,3 ]
saludar( nombre = "Ana" )

# ✅ Bien
total = precio + iva * cantidad
lista = [1, 2, 3]
saludar(nombre="Ana")
```

Las dos versiones hacen **exactamente lo mismo**, pero la segunda se lee sin esfuerzo.

## 5. Imports

Las reglas de la PEP 8 para los `import`:

- Van **al principio del archivo**, justo después del docstring del módulo.
- **Uno por línea** (`import os` y `import sys` en líneas separadas, no `import os, sys`).
- Agrupados en **tres bloques**, separados por una línea en blanco y en este orden:
  1. Librería **estándar** (`os`, `sys`, `datetime`...).
  2. Librerías de **terceros** (las que instalas con `uv add`: `requests`, `pytest`...).
  3. Tus **propios módulos** (los del proyecto).

```python
# 1) Estándar
import os
from datetime import datetime

# 2) Terceros
import requests

# 3) Propios
from cuenta import Cuenta
```

> 💡 Evita `from modulo import *`: trae nombres "a ciegas" y hace imposible saber de
> dónde viene cada cosa. Importa solo lo que necesitas. (Ordenar y agrupar imports
> también lo hace Ruff automáticamente.)

## 6. Comentarios y docstrings

Hay **dos** formas de documentar, y no son lo mismo:

- **Comentarios** (`#`): explican *por qué* se hace algo, no *qué* hace (eso debería
  verse en el propio código). Van con un espacio tras la almohadilla: `# comentario`.
- **Docstrings** (`"""..."""`): cadenas al principio de un módulo, función o clase que
  describen **para qué sirven**. Son accesibles en tiempo de ejecución y las usan las
  herramientas de ayuda.

````python
def calcular_precio_final(precio, descuento):
    """Devuelve el precio tras aplicar un descuento (entre 0 y 1)."""
    # aplicamos un mínimo de 0 por si el descuento es mayor que el precio
    return max(0, precio - precio * descuento)


print(calcular_precio_final(100, 0.2))

# El docstring queda accesible en tiempo de ejecución:
print(calcular_precio_final.__doc__)
````

> 💡 Un buen comentario explica lo que el código **no** puede decir por sí mismo (un
> *porqué*, una decisión, una advertencia). Un mal comentario repite lo evidente:
> `i += 1  # sumamos uno a i` no aporta nada.

## 7. Convenciones idiomáticas ("código pythonico")

Más allá del formato, la PEP 8 recomienda ciertas formas de escribir que se consideran
"de Python". Dos muy habituales:

**Comparar con `None` usando `is`**, no `==`:

````python
valor = None

if valor is None:  # ✅ forma recomendada
    print("no hay valor")

# 'valor == None' funciona, pero el estilo correcto es 'is None'
````

**Aprovechar la "veracidad" (*truthiness*)** de los objetos en vez de comparaciones
redundantes: una lista vacía, una cadena vacía o el `0` ya son "falsos" de por sí.

````python
lista = []

# ✅ pythonico
if not lista:
    print("la lista está vacía")

# ⚠️ redundante: if len(lista) == 0:  /  if lista == []:

nombre = "Ana"
if nombre:  # ✅ 'si hay nombre' (cadena no vacía)
    print(f"Hola, {nombre}")
````

---

## 8. Las reglas son una guía, y las herramientas las aplican

Dos ideas para cerrar:

- **PEP 8 no es un dogma.** Su propio texto dice que la legibilidad manda y que, a
  veces, romper una regla está justificado si el resultado se lee mejor. Lo importante
  es ser **consistente** dentro de un proyecto.
- **No lo apliques a mano.** Como vimos en el vídeo de Ruff, las herramientas
  comprueban y arreglan casi todo esto por ti:

```bash
uv run ruff check    # detecta problemas de estilo (y muchos más)
uv run ruff format   # reescribe el código siguiendo el estilo
```

Conocer las reglas te ayuda a **entender** lo que hacen esas herramientas y a escribir
buen código incluso donde ellas no llegan (como elegir buenos nombres, que ninguna
máquina puede hacer por ti).

---

## 9. Resumen

La PEP 8 es la guía de estilo de Python, y su objetivo es la **legibilidad** mediante
la **consistencia**:

1. **Nombres**: `snake_case` para variables y funciones, `PascalCase` para clases,
   `MAYUSCULAS` para constantes, y nombres **descriptivos**.
2. **Disposición**: 4 espacios de indentación, límite de línea (79 en PEP 8, 88 con
   Ruff/Black), y líneas en blanco para separar (2 entre elementos top-level).
3. **Espacios**: alrededor de operadores y tras las comas; no por dentro de los
   paréntesis.
4. **Imports**: al principio, uno por línea, en tres bloques (estándar, terceros,
   propios).
5. **Comentarios y docstrings**: explican el *porqué* y el *para qué*, no lo obvio.
6. **Estilo idiomático**: `is None`, aprovechar la *truthiness*, etc.
7. Las reglas son una **guía** (la legibilidad manda) y herramientas como **Ruff** las
   aplican por ti.

| Elemento | Convención PEP 8 |
|----------|------------------|
| Variable / función | `snake_case` |
| Clase | `PascalCase` |
| Constante | `MAYUSCULAS_CON_GUION` |
| Indentación | 4 espacios |
| Longitud de línea | 79 (PEP 8) / 88 (Ruff, Black) |
| Imports | Arriba, uno por línea, en 3 bloques |
| Comparar con None | `is None` |

Escribir siguiendo PEP 8 hace que tu código se vea **profesional** y que cualquier
persona que sepa Python pueda leerlo sin esfuerzo.
