# Clases, objetos y estado

En los vídeos anteriores montamos nuestro entorno de trabajo, aprendimos a gestionar
proyectos con `uv` y a cuidar la calidad del código con Ruff. A partir de aquí damos
un paso importante: empezamos con la **Programación Orientada a Objetos (POO)**.

La POO es una forma de organizar nuestro código que nos permite juntar, bajo un
mismo concepto, los **datos** y el **comportamiento** que va con ellos. Es la base
sobre la que se construye casi todo el Python "de verdad" que te vas a encontrar.

En este primer vídeo veremos los cimientos:

- Qué es una **clase** y qué es un **objeto** (o instancia).
- Cómo crear objetos a partir de una clase.
- Qué hacen `__init__` y `self`.
- La diferencia entre **atributos de instancia** y **atributos de clase**.

> 📚 Documentación oficial de referencia:
> **[Tutorial de clases de Python](https://docs.python.org/es/3/tutorial/classes.html)**.

---

## 1. El problema: datos sueltos por todas partes

Imagina que queremos representar a un **usuario** de nuestra aplicación. Con lo que
sabemos hasta ahora, lo más natural sería usar variables sueltas o un diccionario:

```python
# Opción A: variables sueltas
nombre = "Ana"
edad = 25

# Opción B: un diccionario
usuario = {"nombre": "Ana", "edad": 25}

print(usuario["nombre"], usuario["edad"])
```

Esto funciona, pero tiene problemas en cuanto la aplicación crece:

- Si tenemos **muchos** usuarios, acabamos con un montón de diccionarios sueltos y
  ninguna garantía de que todos tengan las mismas claves.
- El **comportamiento** (por ejemplo, "saludar" o "comprobar si es mayor de edad")
  vive en funciones separadas, lejos de los datos a los que pertenece.
- Nadie nos impide escribir `usuario["nombe"]` (con un typo) y reventar en tiempo de
  ejecución.

La POO nos da una herramienta para juntar **datos + comportamiento** en un solo sitio:
la **clase**.

---

## 2. Clase vs objeto: el molde y la pieza

La idea central de la POO se entiende con una metáfora:

- Una **clase** es un **molde**: describe *cómo son* las cosas de un cierto tipo, pero
  no es una cosa concreta.
- Un **objeto** (o **instancia**) es una **pieza concreta** fabricada con ese molde.

Por ejemplo, la clase `Usuario` describe "qué es ser un usuario", y `Ana` o `Luis`
serían objetos concretos creados a partir de esa clase.

Vamos con la clase más simple posible:

```python
class Usuario:
    pass  # de momento no hace nada


# Crear un objeto = "instanciar" la clase. Fíjate en los paréntesis:
u = Usuario()

print(u)
print(type(u))
```

Dos detalles importantes de lo que acabas de ver:

- `class Usuario:` **define** el molde. Por convención, los nombres de clase se
  escriben en `PascalCase`: `Usuario`, `CuentaBancaria`...
- `Usuario()` (con paréntesis) **crea una instancia**. Esto es lo que de verdad
  produce un objeto con el que trabajar.

> 💡 No confundas **definir** la clase (`Usuario`) con **instanciarla** (`Usuario()`).
> Lo primero crea el molde una sola vez; lo segundo fabrica una pieza nueva cada vez
> que lo llamas.

---

## 3. El constructor `__init__` y el famoso `self`

Una clase vacía no es muy útil. Lo normal es que, al crear un objeto, queramos darle
unos **datos iniciales** (su nombre, su edad...). De eso se encarga un método
especial llamado `__init__`, el **constructor**.

`__init__` se ejecuta **automáticamente** cada vez que creamos una instancia, y es
donde inicializamos el **estado** del objeto.

```python
class Usuario:
    def __init__(self, nombre, edad):
        self.nombre = nombre  # guardamos los datos en el objeto
        self.edad = edad


ana = Usuario("Ana", 25)

print(ana.nombre)
print(ana.edad)
```

### Añadir comportamiento: los métodos

Aquí está la gran ventaja frente al diccionario: podemos guardar **comportamiento**
junto a los datos. Un **método** no es más que una función definida dentro de la
clase, y su primer parámetro siempre es `self`:

```python
class Usuario:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        return f"Hola, soy {self.nombre} y tengo {self.edad} años"

    def es_mayor_de_edad(self):
        return self.edad >= 18


ana = Usuario("Ana", 25)
print(ana.saludar())
print(ana.es_mayor_de_edad())
```

Fíjate en que llamamos al método con `ana.saludar()` **sin pasar `self`**: Python lo
rellena por nosotros con el objeto que está a la izquierda del punto (`ana`).

> 💡 Dentro de un método, accedemos al estado del objeto siempre a través de `self`
> (`self.nombre`, `self.edad`). Si te olvidas del `self`, Python pensará que te
> refieres a una variable local y dará un error.

---

## 4. Cada objeto tiene su propio estado

Una de las ideas más importantes: cada instancia guarda **sus propios datos**. Dos
usuarios distintos son independientes entre sí.

```python
ana = Usuario("Ana", 25)
luis = Usuario("Luis", 30)

print(ana.saludar())
print(luis.saludar())

# Cambiar el estado de uno NO afecta al otro
ana.edad = 26
print(ana.edad)   # 26
print(luis.edad)  # 30, intacto
```

A esto nos referimos cuando hablamos de **estado**: el conjunto de datos que un
objeto lleva dentro en un momento dado (`nombre`, `edad`...). El estado puede cambiar
a lo largo de la vida del objeto, y cada instancia tiene el suyo.

---

## 5. Atributos de instancia vs atributos de clase

Hasta ahora todos nuestros atributos vivían en `self` dentro de `__init__`: son
**atributos de instancia** (propios de cada objeto).

Pero también podemos definir atributos directamente en el **cuerpo de la clase**. Esos
son **atributos de clase** y se **comparten** entre todas las instancias. Son útiles
para valores comunes a todos los objetos.

```python
class Usuario:
    idioma = "es"  # atributo de CLASE: compartido por todos los usuarios

    def __init__(self, nombre, edad):
        self.nombre = nombre  # atributos de INSTANCIA: propios de cada uno
        self.edad = edad


ana = Usuario("Ana", 25)
luis = Usuario("Luis", 30)

# El nombre es propio de cada instancia...
print(ana.nombre, luis.nombre)

# ...pero el idioma es el mismo, viene de la clase
print(ana.idioma, luis.idioma)
```

Resumido en una tabla:

| | Atributo de instancia | Atributo de clase |
|---|---|---|
| **Dónde se define** | En `__init__`, sobre `self` | En el cuerpo de la clase |
| **A quién pertenece** | A cada objeto por separado | A la clase (compartido) |
| **Ejemplo** | `self.nombre = nombre` | `idioma = "es"` |
| **Para qué** | Datos que cambian de un objeto a otro | Valores comunes a todos |

Podemos acceder al atributo de clase tanto desde una instancia (`ana.idioma`) como
desde la propia clase (`Usuario.idioma`):

```python
print(Usuario.idioma)  # acceso a través de la clase
print(ana.idioma)      # acceso a través de una instancia
```

> 💡 **Atención**, esto es importante: los atributos de clase parecen muy cómodos,
> pero esconden una **trampa peligrosa** cuando el valor compartido es *mutable*
> (por ejemplo, una lista). Eso es exactamente lo que veremos en el **próximo vídeo**,
> porque es una de las fuentes de bugs más comunes (y desconcertantes) al empezar con
> POO.

---

## 6. Resumen

En este vídeo hemos puesto los cimientos de la POO:

1. Una **clase** es un molde que describe un tipo de objeto; un **objeto** (o
   **instancia**) es una pieza concreta creada con ese molde (`Usuario()`).
2. El **constructor** `__init__` se ejecuta al crear el objeto y sirve para
   inicializar su **estado**.
3. `self` es **el propio objeto**: a través de él guardamos y leemos su estado
   (`self.nombre`) y definimos su comportamiento (los **métodos**).
4. Cada instancia tiene sus **atributos de instancia** propios, mientras que los
   **atributos de clase** se comparten entre todas.

| Concepto | Qué es | Ejemplo |
|----------|--------|---------|
| Clase | El molde | `class Usuario:` |
| Instancia / objeto | Una pieza concreta | `ana = Usuario("Ana", 25)` |
| Constructor | Se ejecuta al crear el objeto | `def __init__(self, ...):` |
| `self` | El propio objeto | `self.nombre = nombre` |
| Método | Comportamiento del objeto | `def saludar(self):` |
| Atributo de instancia | Dato propio de cada objeto | `self.nombre` |
| Atributo de clase | Dato compartido | `idioma = "es"` |
